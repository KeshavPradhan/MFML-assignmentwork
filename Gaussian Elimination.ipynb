```python
def gaussian_elimination(A, b):
    n = len(A)

    # Form the augmented matrix
    for i in range(n):
        A[i].append(b[i])

    # Forward Elimination
    for i in range(n):
        # Pivot check
        if A[i][i] == 0:
            for j in range(i + 1, n):
                if A[j][i] != 0:
                    A[i], A[j] = A[j], A[i]
                    break

        pivot = A[i][i]
        for k in range(i, n + 1):
            A[i][k] /= pivot

        for j in range(i + 1, n):
            factor = A[j][i]
            for k in range(i, n + 1):
                A[j][k] -= factor * A[i][k]

    # Back substitution
    x = [0] * n
    for i in range(n - 1, -1, -1):
        x[i] = A[i][n]
        for j in range(i + 1, n):
            x[i] -= A[i][j] * x[j]

    return x

n = int(input("Enter number of variables: "))

print("\nEnter the coefficients matrix A row by row:")
A = []
for i in range(n):
    row = list(map(float, input(f"Row {i+1}: ").split()))
    A.append(row)

print("\nEnter the constant vector b:")
b = []
for i in range(n):
    val = float(input(f"b[{i+1}]: "))
    b.append(val)

solution = gaussian_elimination(A, b)

print("\nSolution:")
for i in range(n):
    print(f"x{i+1} = {solution[i]}")
```
